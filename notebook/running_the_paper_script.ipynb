```python
rng = prime_rng(137)

p = rng.randprime(1000, 10000)
q = rng.randprime(1000, 10000)

print(p, q)
```

```python
n = p*q
print(n)
```

```python
phi = (p-1) * (q-1)
print(phi)
```

```python
from math import lcm
lcm((p-1), (q-1))
```

```python
lamb = carmichael(p, q)
print(lamb)
```

```python
d1 = inverse257(phi)
print(d1)
```

```python
d2 = inverse257(lamb)
print(d2)
```

```python
e = 257
secrets = [p, q, n, lamb, e, d1, d2]

for num in secrets:
    bitlength = len(str(bin(num)))-2
    print(f"{bitlength:3.0f}    {num}")
```
