```python
def powmod(base: int, exponent: int, modulus: int) -> int:
    out = 1
    pow = base
    for bit in str(bin(exponent))[2:]:
        print(bit, pow, out)
        if bool(int(bit)):
            out = (out * pow) % modulus
        pow = (pow << 1) % modulus 
```
